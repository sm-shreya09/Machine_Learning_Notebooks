# **OPENING CSV FILE FROM URL**

```python
#code snipent - server s csv file ka data load krna ho

import requests
from io import StringIO

# Replace this with the actual URL of your CSV file.
# Using a sample public CSV URL for demonstration.
url = "https://raw.githubusercontent.com/plotly/datasets/master/iris-data.csv"

headers = {"User-Agent": "Mozilla/S.0 (Macintosh;Intel Mac Os X 10.14;)Gecmo/20100101 Firefox/66.0"}
req = requests.get(url,headers=headers)
data = StringIO(req.text)

df_from_url = pd.read_csv(data)
display(df_from_url.head())
```

#**HEADERPARAMETER**

```python
pd.read_csv('aug_train.csv' , header =0)  # when columns are treated as rows not in this case but sometimes it happens ki the 0 can be the cols names but is in the form of row
#to make it a column name we use this parameter
```

#**USE COLS PARAMETER**

```python
pd.read_csv('aug_train.csv' , usecols=['id' ,'Gender', 'Age']) #we can print or decide which column can be printed and which not
```

#**squeeze parameters**

```python
pd.read_csv('aug_train.csv' , usecols =['Gender'])['Gender']
```

#**skiprows/nrowa parameter**

```python
pd.read_csv('aug_train.csv' ,skiprows=[0,4]) #skips the mentioned rows
```

```python
pd.read_csv('aug_train.csv' ,nrows=100) #restricts the number of rows being printed
```

# **encoding parameter**

```python
pd.read_csv('aug_train.csv' , encoding ='latin-1') #sometimsit may happen ki encoding of the charcters are not in UTF 8 to specify that we use encoding parameter
```

#**skip bad lines**

```python
pd.read_csv('aug_train.csv' , on_bad_lines='skip') #skips the rows with multiple values
```

# **dtypes parameter**

```python
pd.read_csv('aug_train.csv', dtype={'Annual_Premium': int}) #the datatype of the Annual_Premium has been changed to integer
```

# **converters**

```python
# Load the CSV file
df = pd.read_csv('aug_train.csv')

# Rename the 'Driving_License' column to 'DL'
df = df.rename(columns={'Driving_License': 'DL'})

# Display the DataFrame to see the change
display(df.head())
```

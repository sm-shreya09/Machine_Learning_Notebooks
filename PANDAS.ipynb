```python
#Joins --> left right outer inner
import pandas as pd
department_info = {

                    "Dept" : ["HR", "IT", "Finance"],
                    "Location": ["NewYork", "San Francisco", "Chicago"],
                    "Manager": ["Alice", "Bob", "Charlie"]
                    }

df2 = pd.DataFrame(department_info)
df2









```

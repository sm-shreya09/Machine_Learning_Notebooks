### Questions for Data Understanding:

1.  **What does each row represent?** (e.g., a single transaction, a customer, a day's sales)
2.  **What do the columns mean?** (Understand the definition and units of each feature).
3.  **What is the overall shape and size of the dataset?** (Number of rows, columns).
4.  **Are there any missing values?** If so, how many, where are they, and what strategies can be used to handle them (e.g., imputation, removal)?
5.  **What are the data types of each column?** Are they appropriate (e.g., dates as datetime objects, categories as categorical)?
6.  **What are the descriptive statistics for numerical columns?** (Mean, median, min, max, standard deviation, quartiles). This helps identify outliers or unusual distributions.
7.  **What are the unique values and their counts for categorical columns?** Are there too many unique values (high cardinality)?
8.  **How is the data distributed for key features?** (Histograms, box plots).
9.  **Are there any outliers or anomalies that need investigation?**
10. **What are the relationships between different variables?** (Correlation matrices, scatter plots for numerical; group-by analyses for categorical/numerical).
11. **Is there any temporal aspect to the data?** If so, what is the time range, frequency, and are there any trends or seasonality?
12. **What is the quality of the data?** Are there inconsistencies, errors, or duplicates?
13. **What is the target variable (if any) and its distribution?**

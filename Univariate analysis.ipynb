# **Univariate analysis **
is a statistical method used to describe the characteristics of a single variable. It involves examining the distribution, central tendency (mean, median, mode), and dispersion (range, variance, standard deviation) of a single feature within a dataset.

-single variable analysis is UniVariateAnalysis


-data with multiple cols ..eacch col is var and independent analysis of each col is univariate analysis.

type of data-Numerical
            - Categorical

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('train.csv') #susrvived =0 means lost there life and 1 means alive
```

```python
df.head()
```

# **1.CATEGORICAL DATA**


## **countplot**

```python
plt.figure(figsize=(6, 4)) # Adjust figure size for better readability
sns.countplot(x='Survived', data=df) # Specify x-axis clearly
plt.title('Count of Survivors (0=No, 1=Yes)') # Add a clear title
plt.xlabel('Survived')
plt.ylabel('Count')
plt.xticks(ticks=[0, 1], labels=['Not Survived', 'Survived']) # Set custom x-axis labels
plt.show()
```

```python
#sns.countplot(df['Survived'])
df['Survived'].value_counts().plot(kind='bar')
```

```python
df['Pclass'].value_counts().plot(kind='bar')
```

## **PieChart**

```python
df['Sex'].value_counts().plot(kind='pie' , autopct = '%1.0f%%')
```

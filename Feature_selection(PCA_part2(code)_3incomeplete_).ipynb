```python
import numpy as np
import pandas as pd

np.random.seed(23)

mu_vec1= np.array([0,0,0])
cov_mat1 = np.array([[1,0,0], [0,1,0], [0,0,1]])
class1_sample = np.random.multivariate_normal(mu_vec1, cov_mat1, 20)
print('class1_sample: \n', class1_sample)
```

```python
df = pd.DataFrame(class1_sample, columns=['feature1', 'feature2', 'feature3'])
df['target'] =1

mu_vec2 = np.array([1,1,1])
cov_mat2= np.array([[1,0,0],[0,1,0], [0,0,1]])
class2_sample = np.random.multivariate_normal(mu_vec2, cov_mat2, 20)
print('class2_sample: \n', class2_sample)
```

```python
df1 = pd.DataFrame(class2_sample, columns=['feature1', 'feature2', 'feature3'])
df1['target'] = 2
df = pd.concat([df, df1], ignore_index=True)
df = df.sample(40)
```

```python
import plotly.express as px
#y_train = y_train.astype(str)
fig= px.scatter_3d(df, x='feature1', y='feature2', z='feature3', color='target')
fig.update_traces(marker_size=5)
fig.show()
```

```python
 #step 1 Apply standard scaling
 from sklearn.preprocessing import StandardScaler
 scaler = StandardScaler()
 df_scaled = scaler.fit_transform(df.drop('target', axis=1))
 df_scaled
```

```python
df.iloc[:, :3] = scaler.fit_transform(df.iloc[:, :3])
```

```python
#Step2 - find covariance matrix
covariance_matrix = np.cov([df.iloc[:,0], df.iloc[:,1], df.iloc[:,2]])
print('covariance matrix:\n', covariance_matrix)
```

```python

#Step3 - find eigen values and eigen vectors

eigen_values , eigen_vectors = np.linalg.eig(covariance_matrix)
print('eigen values:\n', eigen_values)
print('eigen vectors:\n', eigen_vectors)
```

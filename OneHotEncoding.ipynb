# **OneHotEncoding with TopCategories**

```python
df['brand'].value_counts()
```

```python
counts = df['brand'].value_counts()
```

```python
df['brand'].nunique()
threshold = 100
```

```python
repl = counts[counts<= threshold].index
```

```python
pd.get_dummies(df['brand'].replace(repl, 'uncommon').sample(5))
```

```python
# Apply the 'uncommon' replacement to the 'brand' column in X_train and X_test
X_train['brand'] = X_train['brand'].replace(repl, 'uncommon')
X_test['brand'] = X_test['brand'].replace(repl, 'uncommon')
```

```python
# One-hot encode the 'brand' column
ohe_brand = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
X_train_brand_ohe = ohe_brand.fit_transform(X_train[['brand']])
X_test_brand_ohe = ohe_brand.transform(X_test[['brand']])
```

```python
# Reconstruct X_train_final and X_test_final with km_driven and all one-hot encoded features
X_train_final = np.concatenate((X_train[['km_driven']].values, X_train_new, X_train_brand_ohe), axis=1)
X_test_final = np.concatenate((X_test[['km_driven']].values, X_test_new, X_test_brand_ohe), axis=1)

print('Shape of X_train_final:', X_train_final.shape)
print('Shape of X_test_final:', X_test_final.shape)
```

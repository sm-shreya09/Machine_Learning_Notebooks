Process of Fetching API Data in Python
The process shown in your notebook fetches data from a web server (TMDB API) into Python using the requests library in four main steps:

Importing the Library: import requests loads Python's built-in tool for making HTTP network calls.

Sending an HTTP GET Request: requests.get('URL') sends a request to the server at the specified URL endpoint, passing query parameters such as your api_key and language.

Receiving the Response: The server processes the request and sends back a response object containing a Status Code (e.g., <Response [200]>), response headers, and raw payload data.

Parsing the Data: By default, requests.get() returns a response object rather than the raw data. To read the returned dataset in JSON format, save the response to a variable and call the .json() method:

Python

Common HTTP Status Codes
HTTP status codes are three-digit numbers sent by a server to indicate the status of a request:

2xx Success (Request succeeded)
200 OK: The request was successful, and the requested data was returned.

201 Created: The request succeeded and resulted in a new resource being created (common for POST requests).

204 No Content: The request succeeded, but there is no payload returned in the response body.

4xx Client Errors (Issue with the request sent)
400 Bad Request: The server could not understand the request due to invalid syntax or missing required parameters.

401 Unauthorized: Authentication failed or was not provided (e.g., missing or invalid api_key).

403 Forbidden: You are authenticated, but you do not have permission to access the requested resource.

404 Not Found: The endpoint URL or requested resource does not exist on the server.

429 Too Many Requests: You have exceeded the rate limit (sent too many requests in a given time frame).

5xx Server Errors (Issue on the provider's server)
500 Internal Server Error: The server encountered an unexpected error while processing the request.

502 Bad Gateway: The server received an invalid response from an upstream server.

503 Service Unavailable: The server is temporarily overloaded or down for maintenance.

```python
all_dfs = []
for i in range (1 ,127):
  # Correctly format the URL to include the page number
  response = requests.get(f'https://api.themoviedb.org/3/tv/top_rated?api_key=40ce6d2d68786e189606e98b6c89eb07&language=en-US&page={i}')
  if response.status_code == 200:
      temp_df = pd.DataFrame(response.json()['results'])[['id', 'name', 'original_name' , 'vote_average' ]]
      all_dfs.append(temp_df) # Append to a list
  else:
      print(f"Error fetching page {i}: {response.status_code}")

# Concatenate all DataFrames in the list after the loop
df = pd.concat(all_dfs, ignore_index=True)
df
```
